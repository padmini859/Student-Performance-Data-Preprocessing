## Checking Missing Values


```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
numerical_columns = [
    'StudentID',
    'AttendanceRate',
    'StudyHoursPerWeek',
    'PreviousGrade',
    'ExtracurricularActivities',
    'FinalGrade',
    'Study Hours',
    'Attendance (%)'
]

categorical_columns = [
    'Name',
    'Gender',
    'ParentalSupport',
    'Online Classes Taken'
]

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)
```

```python
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

print("Numerical missing values handled successfully!")
```

```python
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0]).infer_objects(copy=False)

print("Categorical missing values handled successfully!")
```

```python
print("Missing values after handling:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

##Checking and Handling Duplicates

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate rows after removal:")
print(df.duplicated().sum())
```

## Encoding Categorical Data

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

for column in categorical_columns:
    df[column] = label_encoder.fit_transform(df[column].astype(str))

print("Categorical data encoded successfully!")
```

```python
print(df.head())
```

##Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaling_columns = [
    'AttendanceRate',
    'StudyHoursPerWeek',
    'PreviousGrade',
    'ExtracurricularActivities',
    'FinalGrade',
    'Study Hours',
    'Attendance (%)'
]

scaler = StandardScaler()

df[scaling_columns] = scaler.fit_transform(df[scaling_columns])

print("Feature scaling completed successfully!")
```

```python
print("Dataset after feature scaling:")
print(df.head())
```
